```python

import pandas as pd
import scipy as sp
```

```python

# todo: comment the code

thresholds = []
aa_effects_mean = []
ab_effects_mean = []
t1_percents = []
t2_percents = []

for threshold in range(20, 50):

    a1_means = []
    a2_means = []
    b_means = []
    aa_deltas = []
    ab_deltas = []
    aa_effects = []
    ab_effects = []
    t1_errors = []
    t2_errors = []
    n = 1000

    for i in range(n):
        # Generate normal samples
        a1 = sp.stats.norm.rvs(loc=1000, scale=100, size=50, random_state=None)
        a2 = sp.stats.norm.rvs(loc=1000, scale=100, size=50, random_state=None)
        b = sp.stats.norm.rvs(loc=1000, scale=100, size=50, random_state=None)

        # add effect to group b
        b += 50

        #
        a1_mean = a1.mean()
        a2_mean = a2.mean()
        b_mean = b.mean()
        aa_delta = abs(a1_mean - a2_mean)
        ab_delta = abs(a1_mean - b_mean)
        aa_effect = aa_delta / a1_mean
        ab_effect = ab_delta / a1_mean
        t1 = int(aa_delta > threshold)
        t2 = int(ab_delta < threshold)

        a1_means.append(a1_mean)
        a2_means.append(a2_mean)
        b_means.append(b_mean)
        aa_deltas.append(aa_delta)
        ab_deltas.append(ab_delta)
        aa_effects.append(aa_effect)
        ab_effects.append(ab_effect)
        t1_errors.append(t1)
        t2_errors.append(t2)

    df = pd.DataFrame(
        {
            'a1_means' : a1_means,
            'a2_means' : a2_means,
            'b_means' : b_means,
            'aa_deltas' : aa_deltas,
            'ab_deltas' : ab_deltas,
            'aa_effects' : aa_effects,
            'ab_effects' : ab_effects,
            't1_errors' : t1_errors,
            't2_errors' : t2_errors
        }
    )

    thresholds.append(threshold)
    aa_effects_mean.append(df.aa_effects.mean())
    ab_effects_mean.append(df.ab_effects.mean())
    t1_percents.append(df.t1_errors.sum() / n)
    t2_percents.append(df.t2_errors.sum() / n)


df2 = pd.DataFrame(
    {
        'thresholds' : thresholds,
        'aa_effects_mean' : aa_effects_mean,
        'ab_effects_mean' : ab_effects_mean,
        't1_percents' : t1_percents,
        't2_percents' : t2_percents
    }
)

df2
```
