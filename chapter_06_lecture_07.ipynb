```python
import numpy as np
import pandas as pd
import scipy as sp
import matplotlib.pyplot as plt
from datetime import datetime
```

```python
def get_min_n(alpha, beta, a, b, e):
    ppf1 = sp.stats.norm.ppf(1 - alpha / 2)
    ppf2 = sp.stats.norm.ppf(1 - beta)
    n = (ppf1 + ppf2)**2 * (a.var() + b.var()) / e**2
    return np.ceil(n)
```

```python
alpha = 0.05
beta = 0.2
e = 0.0025

df = pd.read_csv('data/chapter_06_variance_reduction.csv')
n1 = get_min_n(alpha, beta, df.load_time, df.load_time, e)
n1

df2 = df[(df.load_time < 1)]
n2 = get_min_n(alpha, beta, df2.load_time, df2.load_time, e)
n2

df3 = df[(df.load_time > 0) & (df.load_time < 1)]
n3 = get_min_n(alpha, beta, df3.load_time, df3.load_time, e)
n3

n1, n2, n3
```
